## 13.1 Write the current date as a string to the text file today.txt.

```python
from datetime import date

now = date.today()
now_str = now.isoformat()

with open("today.txt", "w") as file:
    file.write(now_str)

print("The current date was written to today.txt:", now_str)
```

## 13.2 Read the text file today.txt into the string today_string.

```python
with open("today.txt", "r") as file:
    today_string = file.read()

print("today_string =", today_string)
```

## 13.3 Parse the date from today_string.

```python
from datetime import datetime

parsed_date = datetime.strptime(today_string, "%Y-%m-%d").date()

print("Parsed date:", parsed_date)
print("Year:", parsed_date.year)
print("Month:", parsed_date.month)
print("Day:", parsed_date.day)
```

## 15.1 Use multiprocessing to create three separate processes.

Each process waits a random number of seconds between 0 and 1, prints the current time, and then exits.

```python
import subprocess
import sys
import textwrap

code = """
import multiprocessing
import time
import random
from datetime import datetime

def process_task(process_number):
    wait_time = random.uniform(0, 1)
    time.sleep(wait_time)
    print(f"Process {process_number} current time: {datetime.now().strftime('%H:%M:%S')} | Waited: {round(wait_time, 2)} seconds")

if __name__ == "__main__":
    processes = []

    for i in range(3):
        p = multiprocessing.Process(target=process_task, args=(i + 1,))
        processes.append(p)
        p.start()

    for p in processes:
        p.join()

    print("All processes have finished.")
"""

with open("multiprocessing_assignment.py", "w") as file:
    file.write(code)

result = subprocess.run(
    [sys.executable, "multiprocessing_assignment.py"],
    capture_output=True,
    text=True
)

print(result.stdout)

if result.stderr:
    print(result.stderr)
```
